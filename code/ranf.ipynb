```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
import matplotlib.pyplot as plt

data = pd.read_csv('../heart_statlog_cleveland_hungary_final.csv')

print("First 5 rows of data:")
data.head()
```

```python
print("\nSummary Statistics:")
print(data.describe())
print("\nData Structure:")
print(data.info())
```

```python
print(f"\nOriginal shape: {data.shape}")
duplicated_rows = data[data.duplicated()]
print(f"Duplicate rows found: {len(duplicated_rows)}")
data = data.drop_duplicates().reset_index(drop=True)
print(f"Shape after removing duplicates: {data.shape}")
```

```python
print("\nMissing values per column:")
print(data.isnull().sum())
```

```python
print("Missing values per column:")
print(data.isnull().sum())

# Numerical → mean
data.fillna(data.mean(numeric_only=True), inplace=True)

# Categorical → mode
data.fillna(data.mode().iloc[0], inplace=True)
```

```python
categorical_columns = [
    'sex',
    'chest_pain_type',
    'fasting_blood_sugar',
    'resting_ecg',
    'exercise_angina',
    'ST_slope'
]

data = pd.get_dummies(data, columns=categorical_columns, drop_first=True)

print("Shape after encoding:", data.shape)
```

```python
data.head(5)
```

```python
large_scale_features = ['age', 'resting_bp_s', 'cholesterol', 'oldpeak']

for col in large_scale_features:
    
    mu = data[col].mean()      
    sigma = data[col].std()   
    
    data[col] = (data[col] - mu) / sigma

print("After Standardization:")
print(data[large_scale_features].head())
```

```python
numerical_columns = data.select_dtypes(include=['int64', 'float64']).columns
numerical_columns = numerical_columns.drop('target')

for col in numerical_columns:
    
    xmin = data[col].min()
    xmax = data[col].max()
    
    data[col] = (data[col] - xmin) / (xmax - xmin)

print("After Min-Max Normalization:")
data[numerical_columns].head()
```

```python
data.head()
```

```python
X = data.drop("target", axis=1)
y = data["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier

import numpy as np

ranf_model = RandomForestClassifier(
    criterion='gini',
    max_depth=8,
    max_features='sqrt', 
    n_estimators=200,
    random_state=42
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(ranf_model, X_train, y_train, cv=cv, scoring='accuracy')

print(f"CV Scores: {cv_scores}")
print(f"Mean CV Accuracy: {cv_scores.mean()*100:.2f}% (+/- {cv_scores.std()*100:.2f}%)")

ranf_model.fit(X_train, y_train)
y_pred = ranf_model.predict(X_test)
y_proba = ranf_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.model_selection import GridSearchCV
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

param_grid_1 = {
    'n_estimators': [10, 50, 100, 200], 
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None]
}
grid_1 = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid_1,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1
)

grid_1.fit(X_train, y_train)

print("Best Parameters (Recommended Search):")
print(grid_1.best_params_)
print("Best CV Accuracy:", grid_1.best_score_)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix, roc_curve, auc,
    precision_recall_curve, average_precision_score
)
from sklearn.model_selection import learning_curve

y_pred = ranf_model.predict(X_test)
y_proba = ranf_model.predict_proba(X_test)[:, 1]

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Reds',
    ax=axes[0, 0]
)

axes[0, 0].set_title("(a) ranf Confusion Matrix", fontsize=14)
axes[0, 0].set_xlabel("Predicted Label")
axes[0, 0].set_ylabel("True Label")

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

axes[0, 1].plot(fpr, tpr, label=f"AUC = {roc_auc:.2f}")
axes[0, 1].plot([0, 1], [0, 1], 'k--')
axes[0, 1].set_title("(b) ranf ROC AUC Plot", fontsize=14)
axes[0, 1].set_xlabel("False Positive Rate")
axes[0, 1].set_ylabel("True Positive Rate")
axes[0, 1].legend()

# Learning Curve
train_sizes, train_scores, val_scores = learning_curve(
    ranf_model,
    X_train,
    y_train,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

train_mean = np.mean(train_scores, axis=1)
val_mean = np.mean(val_scores, axis=1)
val_std = np.std(val_scores, axis=1)

axes[1, 0].plot(train_sizes, train_mean, 'o-', label="Training Score")
axes[1, 0].plot(train_sizes, val_mean, 'o-', label="Cross Validation Score")
axes[1, 0].fill_between(
    train_sizes,
    val_mean - val_std,
    val_mean + val_std,
    alpha=0.2
)

axes[1, 0].set_title("(c) ranf Learning Curve", fontsize=14)
axes[1, 0].set_xlabel("Training Examples")
axes[1, 0].set_ylabel("Score")
axes[1, 0].legend()

# Precision-Recall Curve
precision, recall, _ = precision_recall_curve(y_test, y_proba)
avg_precision = average_precision_score(y_test, y_proba)

axes[1, 1].plot(recall, precision, label="Binary PR curve")
axes[1, 1].axhline(y=avg_precision, linestyle='--', color='r',
                   label=f"Avg. precision = {avg_precision:.2f}")

axes[1, 1].set_title("(d) ranf Precision-Recall Curve", fontsize=14)
axes[1, 1].set_xlabel("Recall")
axes[1, 1].set_ylabel("Precision")
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

precision_val = precision_score(y_test, y_pred)
recall_val = recall_score(y_test, y_pred)
f1_val = f1_score(y_test, y_pred)
accuracy_val = accuracy_score(y_test, y_pred)

print(f"Precision: {precision_val*100:.2f}%")
print(f"Recall: {recall_val*100:.2f}%")
print(f"F1 Score: {f1_val*100:.2f}%")
print(f"Accuracy: {accuracy_val*100:.2f}%")
print(f"Accuracy Loss: {(1-accuracy_val)*100:.2f}%")
```
